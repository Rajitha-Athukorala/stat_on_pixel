## Putting all the steps together

```python
# Get all the files in a folder

# Center coordinates
# This changes the center of the circle used to crop (So dont change)
center_coordinates = (2940, 1950)
  
# Radius of circle
radius = 1730
   
# Blue color in BGR
color = (255, 0, 0)

# Source path
source = ".\\..\\Data\\third_set"

# Wildcard to use for the image search
wildcard = '/*MED*'

f = []
# Set the wildcard to use for the image search
for files in glob.glob(source + wildcard):
    f.append(files)
    
for files in f:
    image = cv2.imread(files)
    blank = np.zeros(image.shape[:2],dtype='uint8')
    circle = cv2.circle(blank, center_coordinates, radius, color, -1)
    masked = cv2.bitwise_and(image,image,mask=circle)
    file_name = files[:-4]+'_C.JPG'
    cv2.imwrite(file_name, masked)
```
